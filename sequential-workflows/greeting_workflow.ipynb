```python
from typing import TypedDict
```

```python
class GreetingState(TypedDict):
    name: str
    greeting: str
```

```python
def create_greeting(state: GreetingState):
    name =state['name']
    return {'greeting': f"Hi {name}, welcome to langGraph universe."}
```

```python
from langgraph.graph import StateGraph, START, END
```

```python
graph = StateGraph(GreetingState)

graph.add_node("create_greeting",  create_greeting)

graph.add_edge(START,"create_greeting")
graph.add_edge("create_greeting", END)


workflow = graph.compile()
```

```python
result = workflow.invoke({"name": "victor"})
```
