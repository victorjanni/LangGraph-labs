```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class Additionstate(TypedDict):
    input1: int
    input2: int
    result: float
```

```python
def addition(state: Additionstate) -> Additionstate:
    In1 = state['input1']
    In2 = state['input2']
    result = In1 + In2
    state['result'] = result
    return state
```

```python
graph = StateGraph(Additionstate)

graph.add_node('addition', addition)

graph.add_edge(START,'addition')
graph.add_edge('addition',END)

workflow = graph.compile()
```

```python
intial_state = {'input1': 34, 'input2':54}

final_state = workflow.invoke(intial_state)

print(final_state)
```
