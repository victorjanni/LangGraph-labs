```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class SubtracdtionState(TypedDict):
    num1 : int
    num2 : int
    result : float
```

```python
def subtraction(state: SubtracdtionState) -> SubtracdtionState:
    num1 = state['num1']
    num2 = state['num2']
    result = num2 - num1
    state['result'] = result
    return state
```

```python
graph = StateGraph(SubtracdtionState)

graph.add_node('subtraction',subtraction)

graph.add_edge(START,'subtraction')
graph.add_edge('subtraction', END)

workflow = graph.compile()
```

```python
intial_state = {'num1':63, 'num2':45}
final_state = workflow.invoke(intial_state)
print(final_state)
```
